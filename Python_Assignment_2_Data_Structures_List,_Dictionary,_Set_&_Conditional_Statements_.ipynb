**Sets (Operations):**

a. Create a set called my_set with following values:    
['a','e','i','o','u','a','a','i']
Analyse the output and provide explanation for the same.

```python
my_set = {'a','e','i','o','u','a','a','i'}
print(my_set)
```

a) Answer
* Here duplicates are automatically removed.
* Output: {'e', 'i', 'u', 'o', 'a'}, because sets store only unique elements.

```python
# b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.

my_set[4]='s'
print(my_set)
```

b) Answer:
* This error happens because sets are unordered so the indexing will not work in case of sets, so we can't access or change elements by position.

```python
# c. Create two sets:
# set1 with values: {1, 3, 5, 7, 9}
# set2 with values: {2, 3, 5, 8, 10}

set1 = {1,3,5,7,9}
set2 = {2,3,5,8,10}
print(set1)
print(set2)
```

```python
# d. Compute and print the union and intersection of set1 and set2.

print(set1.union(set2))  #Union
print(set1.intersection(set2)) #intersection
```

c) Created two sets set1 and set2 using curly braces to store unique unordered values.

d) print(set1.union(set2)), Computes the union with union() for all distinct elements
* Output: {1, 2, 3, 5, 7, 8, 9, 10}
* print(set1.intersection(set2)), intersection() takes common elements
* Output: {3,5}

**Operators & Conditional Statements :
(IF, ELIF, ELSE)**

Performance Category Program:
1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).
2. Find the performance category based on the input score using following criteria:
a. Above Average:  Score greater than 7
b. Average: Score between 4 and 7(both inclusive)
c. Below Average: Score lesser than 4
3. Output: Print the Performance category
4. Additional Step: You can give a prompt of your choice to each category.  
For eg: If score below average “Need to Improve your performance, consistent
practice will lead to better results”.  

```python
score = int(input("Enter your score(0 to 10): "))
if score >7:
  print("Above Average: Excellent work! Keep it up.")
elif score >=4:
  print("Average: Good effort! Keep practicing, there's little more steps for make excellent.")
else:
  print("Below Average: Need to improve your performance, consistent practice will lead to better results.")
```

This program takes a score from 0 to 10 as input and categorizes performance using if-elif-else. Score >7 prints Above Average, 4 to 7 prints Average and below 4 prints Below Average with a motivational message.
